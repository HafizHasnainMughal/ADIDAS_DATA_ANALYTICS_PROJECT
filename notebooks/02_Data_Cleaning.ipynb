for cleaning process we import the python libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

import the data set

```python
RAW_DATA_PATH="../data/raw/Adidas_Synthetic_Messy_Data_Analytics_Project.xlsx"
df = pd.read_excel(
    RAW_DATA_PATH,
    sheet_name="Messy_Raw_Data"
)
```

confirm the duplicate rows

```python
duplicate_count = df.duplicated().sum()
print(f"the duplicated rows are {duplicate_count}")
```

remove the exact duplicated rows 

```python
rows_before = len(df)
print(f"rows_before length with duplicate rows {rows_before}")
df = df.drop_duplicates(keep="first").reset_index(drop=True)
rows_after = len(df)
print(f"rows after the  length without duplicate rows {rows_after}")
```

```python
df.shape
```

text whitespacing removeing 

```python
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()
for col in text_columns:
    df[col] = df[col].str.strip()
```

country and sale channel standardization

```python
df["Country"]=(df["Country"].str.strip().str.title())
df['Sales_Channel']=(df['Sales_Channel'].str.strip().str.title())

```

invalid valuesin the numerical data type 
steps 
  1 >> range select
  2 >> convert it into the nan
  3 >> fill by  different methods like means ,median ,mode ,knn etc 

```python
df.loc[
    (df["Customer_Age"] < 13) |
    (df["Customer_Age"] > 100),
    "Customer_Age"
] = np.nan

df.loc[
    df["Shipping_Days"] < 0,
    "Shipping_Days"
] = np.nan

df.loc[
    (df["Discount_Rate"] < 0) |
    (df["Discount_Rate"] > 1),
    "Discount_Rate"
] = np.nan

```

```python
from sklearn.impute import KNNImputer
knn = KNNImputer(n_neighbors=5, weights="distance")
```

```python
knn_features = ["Customer_Age","Customer_Rating","Shipping_Days",
                "Discount_Rate","Marketing_Spend_USD","Quantity",
                "Unit_Price_USD","Revenue_USD","Cost_USD","Profit_USD"]
df[knn_features] = knn.fit_transform(
    df[knn_features])
```

```python
df[
    [
        "Customer_Age",
        "Customer_Rating",
        "Shipping_Days",
        "Discount_Rate",
        "Marketing_Spend_USD"
    ]
].describe()
```

```python
missing_percentage = (df.isnull().sum()/ len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_report[
    missing_report["Missing_Values"] > 0].sort_values(by="Missing_Percentage",ascending=False)
```

changing the data type of custumer age,custumer rating ,shiping days

```python
df['Customer_Rating'].unique()
```

changing the customer age,customer rating,shipping days  columns ki data type changes

```python
df['Customer_Age']=df['Customer_Age'].astype(int)
df['Customer_Rating']=df['Customer_Rating'].astype(int)
df['Shipping_Days']=df['Shipping_Days'].astype(int)
```

```python
df['Customer_Age'].unique()
```

```python
df['Discount_Rate'].unique()
```

```python
df['Discount_Rate'] = (df['Discount_Rate'] * 100).round(2)
```

```python
df['Discount_Rate'].unique()
```

```python
df['Marketing_Spend_USD'].unique()
df['Marketing_Spend_USD'].dtypes
```

remaning missing values of categorical data handle

```python
remaining_missing = pd.DataFrame({
""
    "Missing_Count": df.isnull().sum(),

    "Missing_%":
    round(df.isnull().mean()*100,2)
})
remaining_missing = remaining_missing[
    remaining_missing["Missing_Count"]>0
]
remaining_missing
```

```python
invalid = ["?","Unknown","N/A","None"]
df["Gender"] = df["Gender"].replace(invalid, np.nan)
```

```python
df['Gender'].isna().sum()
```

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
df['City'].unique()
```

```python
df["City"] = (
    df.groupby("Country")["City"]
      .transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else np.nan))
)
```

```python
df['Size'].unique()
```

```python
size_mode = df["Size"].mode()[0]
df["Size"] = df["Size"].fillna(size_mode)
```

```python
df['Size'].unique()
```

```python
size_map = {
    "XS": 36,
    "S": 38,
    "M": 40,
    "L": 42,
    "XL": 44
}
df["Size"] = df["Size"].replace(size_map)
df["Size"] = pd.to_numeric(df["Size"], errors="coerce")
```

```python
df["Campaign"].unique()
```

```python
df["Campaign"] = df["Campaign"].fillna("No Campaign")
```

```python
missing_percentage = (df.isnull().sum()/ len(df)) * 100
missing_report = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Missing_Percentage": missing_percentage})
missing_report[
    missing_report["Missing_Values"] > 0].sort_values(by="Missing_Percentage",ascending=False)
```
