create the empty list

```python
cleaning_log = []
```

```python
def log_cleaning_step(
    step,
    description,
    rows_before,
    rows_after,
    notes=""
):
    cleaning_log.append({
        "Step": step,
        "Description": description,
        "Rows_Before": rows_before,
        "Rows_After": rows_after,
        "Rows_Changed": rows_before - rows_after,
        "Notes": notes
    })
```

confirm the duplicate rows

```python
duplicate_count = df.duplicated().sum()
print(f"the duplicated rows are {duplicate_count}")
```

remove the exact duplicated rows 

```python
rows_before = len(df)
print(f"rows_before length with duplicate rows {rows_before}")
df = df.drop_duplicates(keep="first").reset_index(drop=True)
rows_after = len(df)
print(f"rows after the  length without duplicate rows {rows_after}")
```

```python
log_cleaning_step(
    step="Duplicate Removal",
    description="Removed exact duplicate rows",
    rows_before=rows_before,
    rows_after=rows_after,
    notes="Only exact duplicate records were removed."
)
```
