for cleaning process we import the python libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

import the data set

```python
RAW_DATA_PATH="../data/Adidas US Sales Datasets.xlsx"
df = pd.read_excel(RAW_DATA_PATH)
```

copy the raw data for backup

Yeh raw data ki independent copy banata hai.meand original data per koi effect nahi parta

```python
df_raw = df.copy(deep=True)
```

confirm the duplicate rows

```python
duplicate_count = df.duplicated().sum()
print(f"the duplicated rows are {duplicate_count}")
```

remove the exact duplicated rows 

```python
rows_before = len(df)
print(f"rows_before length with duplicate rows {rows_before}")
df = df.drop_duplicates(keep="first").reset_index(drop=True)
rows_after = len(df)
print(f"rows after the  length without duplicate rows {rows_after}")
```

```python
df.shape
```

column renaming


```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)
```

 removeing empty columns

```python
df.drop(columns=["unnamed:_0"], inplace=True, errors="ignore")
```

```python
missing_percentage = (df.isnull().sum()/ len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_report[
    missing_report["Missing_Values"] > 0].sort_values(by="Missing_Percentage",ascending=False)
```

```python
df.to_excel(
    "../data/Adidas_Cleaned_Data.xlsx",
    index=False
)
```
