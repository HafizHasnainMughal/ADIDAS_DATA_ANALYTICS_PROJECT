```python
def customer_value_category(value):

    if value < 500:
        return "Low Value"

    elif value < 2000:
        return "Medium Value"

    else:
        return "High Value"
```

```python
df_analysis["Customer_Value_Category"] = (
    df_analysis[
        "Customer_Lifetime_Value_USD"
    ]
    .apply(customer_value_category)
)
```

```python
df_analysis["Return_Flag"] = (
    df_analysis["Returned"]
    .map({
        "Yes": 1,
        "No": 0
    })
)
```

```python
def discount_category(discount):

    if discount == 0:
        return "No Discount"

    elif discount <= 0.10:
        return "Low Discount"

    elif discount <= 0.30:
        return "Medium Discount"

    else:
        return "High Discount"
```

```python
df_analysis["Discount_Category"] = (
    df_analysis["Discount_Rate"]
    .apply(discount_category)
)
```

```python
df_analysis["Customer_Order_Value"] = (
    df_analysis["Revenue_USD"]
)
```

```python
customer_summary = (
    df_analysis
    .groupby("Customer_ID")
    .agg(
        Total_Revenue=("Revenue_USD", "sum"),
        Total_Orders=("Order_ID", "nunique"),
        Total_Quantity=("Quantity", "sum"),
        Average_Rating=("Customer_Rating", "mean")
    )
    .reset_index()
)
```

```python
customer_frequency = (
    df_analysis
    .groupby("Customer_ID")
    ["Order_ID"]
    .nunique()
    .reset_index(
        name="Purchase_Frequency"
    )
)
```

```python
analysis_date = (
    df_analysis["Order_Date"].max()
    + pd.Timedelta(days=1)
)
```

```python
rfm = (
    df_analysis
    .groupby("Customer_ID")
    .agg(
        Recency=(
            "Order_Date",
            lambda x: (
                analysis_date -
                x.max()
            ).days
        ),

        Frequency=(
            "Order_ID",
            "nunique"
        ),

        Monetary=(
            "Revenue_USD",
            "sum"
        )
    )
    .reset_index()
)
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
)
```

```python
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)
```

```python
rfm.head()
```

custumer segmentation

```python
def rfm_segment(score):

    if score >= 12:
        return "Champions"

    elif score >= 9:
        return "Loyal Customers"

    elif score >= 6:
        return "Potential Customers"

    else:
        return "At Risk"
```

```python
rfm["Customer_Segment_RFM"] = (
    rfm["RFM_Score"]
    .apply(rfm_segment)
)
```
